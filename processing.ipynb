## adding "duration" column
complete_time - request_time

```python
import pandas as pd

df = pd.read_csv("datasets/transfers_filtered.csv")
df["request_time"] = pd.to_datetime(df["request_time"], errors="coerce")
df["complete_time"] = pd.to_datetime(df["complete_time"], errors="coerce")

valid = df.dropna(subset=["request_time", "complete_time"]).copy()
valid["duration_seconds"] = (valid["complete_time"] - valid["request_time"]).dt.total_seconds()

print(valid[["request_time", "complete_time", "duration_seconds"]].head())
```

```python
mask = df["request_time"].notna() & df["complete_time"].notna()
df.loc[mask, "duration_seconds"] = (
    df.loc[mask, "complete_time"] - df.loc[mask, "request_time"]
).dt.total_seconds()
```
